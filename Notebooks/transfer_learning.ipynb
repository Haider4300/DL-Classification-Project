```python
# ============================================
# IMPORTS (Add these at the very top!)
# ============================================
import keras
from keras import layers
import tensorflow as tf
import matplotlib.pyplot as plt

print("✅ All imports successful!")
```

```python
# ============================================
# TRANSFER LEARNING WITH EFFICIENTNET
# ============================================

# Reload data with 224x224 size (EfficientNet requirement)
IMAGE_SIZE_TL = (224, 224)

train_ds_tl = keras.utils.image_dataset_from_directory(
    'data/natural_images',
    image_size=IMAGE_SIZE_TL,
    batch_size=32,
    subset="training",
    validation_split=0.2,
    seed=42
)
```

```python

val_ds_tl = keras.utils.image_dataset_from_directory(
    'data/natural_images',
    image_size=IMAGE_SIZE_TL,
    batch_size=32,
    subset="validation",
    validation_split=0.2,
    seed=42
)
```

```python

# Optimize
train_ds_tl = train_ds_tl.cache().prefetch(tf.data.AUTOTUNE)
val_ds_tl = val_ds_tl.cache().prefetch(tf.data.AUTOTUNE)

# Load pre-trained model
base_model = keras.applications.EfficientNetB3(
    input_shape=(224, 224, 3),
    include_top=False,
    weights='imagenet'
)
base_model.trainable = False
```

```python

# Build model
model_tl = keras.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dense(256, activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(8, activation='softmax')
])

model_tl.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python

# Train
history_tl = model_tl.fit(train_ds_tl, validation_data=val_ds_tl, epochs=15)
```

```python

# Save
model_tl.save("natural_images_transfer_learning.keras")
print("✅ Transfer Learning model saved!")
```
